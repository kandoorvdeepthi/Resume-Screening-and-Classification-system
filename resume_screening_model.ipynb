DATA CLEANING

1 URL
2 hashtages
3 mentions
4 special chars
5 punctuations

```python
import re   #regular expression
```

```python
def cleanresume(txt):
    cleantxt = re.sub("http\S+"," ",txt)
    cleantxt = re.sub("@\S+"," ",cleantxt)
    cleantxt = re.sub("#\S+"," ",cleantxt)
    cleantxt = re.sub('[%s]' % re.escape("""!"#$%&'()*+,-./:;<=>?@[\]^_{|}""") ," ",cleantxt)
    cleantext = re.sub('\n', ' ',cleantxt)
    txt = re.sub(r'[^a-zA-Z\s]', ' ', txt) 
    cleantxt = re.sub("\n+"," ",cleantxt)
    cleantxt = re.sub("\xa0"," ",cleantxt)
    cleantxt = re.sub("\s+"," ",cleantxt )
    cleantxt =cleantxt.strip()
    return cleantxt.lower()
                      
```

```python
df['Resume_str'] = df['Resume_str'].apply(lambda x: cleanresume(x))
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
tfidf = TfidfVectorizer(stop_words = 'english',ngram_range=(1, 2),max_features=50000,sublinear_tf=True, min_df=2   )
tfidf.fit(df['Resume_str'])
requiredtext = tfidf.transform(df['Resume_str'])
```

```python
df
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train,X_test,y_train,y_test = train_test_split(requiredtext,df['Category'],test_size=0.2,random_state=42)
```

```python
X_train.shape   #here 40162 is sparsed matrix which is derived from each word
```

```python
X_test.shape
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.multiclass import OneVsRestClassifier
from sklearn.metrics import accuracy_score
```

```python
clf = OneVsRestClassifier(LogisticRegression(max_iter=1000,C=1.0))
clf.fit(X_train,y_train)
```

```python
ypred = clf.predict(X_test)
print(accuracy_score(y_test,ypred))
```
